```python
import os
import pandas as pd
import numpy as np

raw_dir = "../data/raw"
files = ["courses.csv", "assessments.csv", "vle.csv", "studentInfo.csv", "studentRegistration.csv", "studentAssessment.csv"]
tables = {f.replace(".csv", ""): pd.read_csv(f"{raw_dir}/{f}", na_values=["?"]) for f in files}
for name, df in tables.items():
    print(f"{name}: {len(df):,} rows, {len(df.columns)} cols")
```

### 1. Khảo sát 7 Môn học và Lựa chọn Môn học Trọng tâm

```python
si = tables["studentInfo"]
module_stats = si.groupby(["code_module", "code_presentation"]).agg(
    total_students=("id_student", "count"),
    pass_count=("final_result", lambda s: (s == "Pass").sum()),
    fail_count=("final_result", lambda s: (s == "Fail").sum()),
    withdrawn_count=("final_result", lambda s: (s == "Withdrawn").sum()),
    distinction_count=("final_result", lambda s: (s == "Distinction").sum())
).reset_index()
print(module_stats.head(10))
```

### 2. Định nghĩa Feature Engineering và Target Variable
- Target Variable:  = 1 (Fail/Withdrawn), 0 (Pass/Distinction).
